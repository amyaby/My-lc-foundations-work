## No memory

```python
from langchain.agents import create_agent
from langchain.chat_models import init_chat_model
import os
model = init_chat_model(
    model="gemini-3.1-flash-lite",
    model_provider="google-genai",
    api_key=os.getenv("GOOGLE_API_KEY"),
)
agent = create_agent(
    model=model,
)
```

```python
from langchain.messages import HumanMessage

question = HumanMessage(content="Hello my name is Imane and my favourite colour is purpole")

response = agent.invoke(
    {"messages": [question]} 
)
```

```python
from pprint import pprint

pprint(response)
```

```python
print(response)
```

```python
question = HumanMessage(content="What's my favourite colour?")

response = agent.invoke(
    {"messages": [question]} 
)

pprint(response)
```

----> our agent is _so far_ stateless : since it's output was "I don’t know what your favorite color is yet! Since I don’t have access to your personal history or thoughts, I can’t know unless you tell me"

## With Memory

```python
from langgraph.checkpoint.memory import InMemorySaver

agent = create_agent(
    model=model,
    checkpointer=InMemorySaver(),
)
```

```python
from langchain.messages import HumanMessage

question = HumanMessage(content="Hello my name is Imane and my favourite colour is gray")
config = {"configurable": {"thread_id": "1"}}

response = agent.invoke(
    {"messages": [question]},
    config,  
)
```

```python
pprint(response)
```

```python
print(response)
```

```python
question = HumanMessage(content="What's my favourite colour?")

response = agent.invoke(
    {"messages": [question]},
    config,  
)

pprint(response)
```
